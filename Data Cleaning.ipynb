```python
df['Educational_level'] = df['Educational_level'].fillna('Unknown')
df['Vehicle_driver_relation'] = df['Vehicle_driver_relation'].fillna('Unknown')
df['Driving_experience'] = df['Driving_experience'].fillna('Unknown')
df['Type_of_vehicle'] = df['Type_of_vehicle'].fillna('Unknown')
df['Service_year_of_vehicle'] = df['Service_year_of_vehicle'].fillna('Unknown')
df['Area_accident_occured'] = df['Area_accident_occured'].fillna('Unknown')
df['Lanes_or_Medians'] = df['Lanes_or_Medians'].fillna('Unknown')
df['Road_allignment'] = df['Road_allignment'].fillna('Unknown')
df['Types_of_Junction'] = df['Types_of_Junction'].fillna('Unknown')
df['Road_surface_type'] = df['Road_surface_type'].fillna('Unknown')
df['Vehicle_movement'] = df['Vehicle_movement'].fillna('Unknown')
df['Work_of_casuality'] = df['Work_of_casuality'].fillna('Unknown')
```

```python
df['Owner_of_vehicle'] = df['Owner_of_vehicle'].fillna(
    df['Owner_of_vehicle'].mode()[0]
)

df['Fitness_of_casuality'] = df['Fitness_of_casuality'].fillna(
    df['Fitness_of_casuality'].mode()[0]
)
```

```python
df.head(30)
```

```python
df['Casualty_severity'].unique()
```

```python
df['Casualty_severity'] = df['Casualty_severity'].replace('na', np.nan)
```

```python
df['Casualty_severity'] = df['Casualty_severity'].fillna(
    df['Casualty_severity'].mode()[0]
)
```

```python
df.head(30)
```

```python
df['Sex_of_casualty'] = df['Sex_of_casualty'].replace('na', 'Unknown')
df['Age_band_of_casualty'] = df['Age_band_of_casualty'].replace('na', 'Unknown')
df['Casualty_class'] = df['Casualty_class'].replace('na', 'Unknown')
```
